```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
def vector_rotation_matrix(phi,theta):
    x = [np.cos(phi)*np.cos(theta), np.cos(phi)*np.sin(theta), np.sin(phi)]
    x = x/np.linalg.norm(x)
    
    z = [np.sin(-phi)*np.cos(theta), np.sin(-phi)*np.sin(theta), np.cos(phi)]
    z = z/np.linalg.norm(z)
    
    y = np.cross(z, x)
    y = y/np.linalg.norm(y)
    
    R = np.column_stack([x, y, z])
    
    return R
```

```python
armphiP = -np.pi/4 # inclination
armthetaP = np.pi/4 # azimuth

motorphiP = np.linspace(0, np.pi, 20)             # inclination
motorthetaP = np.linspace(-np.pi/2, np.pi/2, 20)    # azimuth
```

```python
hemisphere = []
vec = [[0, 0, 0]]
vec.append([np.cos(armphiP)*np.cos(armthetaP), np.cos(armphiP)*np.sin(armthetaP), np.sin(armphiP)])
vec = np.array(vec)

for phi in motorphiP:
    for theta in motorthetaP:
        dir = 0.25*np.array([np.sin(phi)*np.cos(theta), np.sin(phi)*np.sin(theta), -np.cos(phi)])
        hemisphere.append(dir)

hemisphere = np.array(hemisphere)

R = vector_rotation_matrix(armphiP, armthetaP)
hemisphere = hemisphere@R.T + vec[1]

x = 0.4*np.array([1, 0, 0])
y = 0.4*np.array([0, 1, 0])
z = 0.4*np.array([0, 0, 1])

x_new = R @ (0.6*x)
y_new = R @ (0.6*y)
z_new = R @ (0.6*z)

x_new += vec[1]
y_new += vec[1]
z_new += vec[1]

print(R)
```

```python
# Create Plotly 3D plot
fig = go.Figure()

# Arm line
fig.add_trace(go.Scatter3d(
    x=vec[:, 0], y=vec[:, 1], z=vec[:, 2],
    mode='lines',
    line=dict(color='black', width=4),
    name='Arm'
))

# Arm endpoint
fig.add_trace(go.Scatter3d(
    x=[vec[1, 0]], y=[vec[1, 1]], z=[vec[1, 2]],
    mode='markers',
    marker=dict(color='black', size=6, symbol='circle'),
    name='Tip'
))

# Hemisphere
fig.add_trace(go.Scatter3d(
    x=hemisphere[:, 0], y=hemisphere[:, 1], z=hemisphere[:, 2],
    mode='markers',
    marker=dict(color='red', size=4),
    name='Motors'
))

# Main axis
fig.add_trace(go.Scatter3d(
    x=[0,x[0]], y=[0,x[1]], z=[0,x[2]],
    mode='lines',
    line=dict(color='red', width=4),
    name='x-axis'
))

fig.add_trace(go.Scatter3d(
    x=[0,y[0]], y=[0,y[1]], z=[0,y[2]],
    mode='lines',
    line=dict(color='blue', width=4),
    name='y-axis'
))

fig.add_trace(go.Scatter3d(
    x=[0,z[0]], y=[0,z[1]], z=[0,z[2]],
    mode='lines',
    line=dict(color='green', width=4),
    name='z-axis'
))

# New axis
fig.add_trace(go.Scatter3d(
    x=[vec[1,0],x_new[0]], y=[vec[1,1],x_new[1]], z=[vec[1,2],x_new[2]],
    mode='lines',
    line=dict(color='red', width=4),
    name='x-axis'
))

fig.add_trace(go.Scatter3d(
    x=[vec[1,0],y_new[0]], y=[vec[1,1],y_new[1]], z=[vec[1,2],y_new[2]],
    mode='lines',
    line=dict(color='blue', width=4),
    name='y-axis'
))

fig.add_trace(go.Scatter3d(
    x=[vec[1,0],z_new[0]], y=[vec[1,1],z_new[1]], z=[vec[1,2],z_new[2]],
    mode='lines',
    line=dict(color='green', width=4),
    name='z-axis'
))

# Layout and Z-axis inversion
fig.update_layout(
    scene=dict(
        xaxis=dict(title='x', range=[-1.5, 1.5]),
        yaxis=dict(title='y', range=[-1.5, 1.5]),
        zaxis=dict(title='z', range=[1.5, -1.5])
    ),
    width=800,
    height=800,
    showlegend=True
)

fig.show()
```
